# Student Marks Analysis

This project analyzes a manually created list of student marks and calculates the total, average, highest, and lowest marks.

**Track:** AI & ML  
**Task:** 1 — Student Marks Analysis

```python
# Student marks stored in a Python list
student_marks = [78, 85, 92, 67, 74, 88, 95, 81, 69, 90]

# Basic calculations
total_marks = sum(student_marks)
average_marks = total_marks / len(student_marks)
highest_marks = max(student_marks)
lowest_marks = min(student_marks)

# Display results
print("Student Marks Analysis")
print("----------------------")
print(f"Number of students : {len(student_marks)}")
print(f"Total marks        : {total_marks}")
print(f"Average marks      : {average_marks:.2f}")
print(f"Highest marks      : {highest_marks}")
print(f"Lowest marks       : {lowest_marks}")
```

## Result

The program uses a Python list to store marks. `sum()` calculates the total, `max()` finds the highest mark, `min()` finds the lowest mark, and the average is calculated as total marks divided by the number of students.
