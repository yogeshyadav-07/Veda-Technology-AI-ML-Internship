Assign grades to students using predefined marks ranges.

## Grading Criteria

- 90–100 → A+
- 80–89 → A
- 70–79 → B
- 60–69 → C
- 50–59 → D
- 40–49 → E
- 0–39 → F
- Outside 0–100 → Invalid

```python
# Student Grade Calculator
# Veda Technology - AI & ML Internship
# Task 2

student_marks = {
    "Aman": 92,
    "Priya": 85,
    "Rahul": 76,
    "Sneha": 68,
    "Vikas": 59,
    "Neha": 43,
    "Karan": 31
}

def calculate_grade(marks):
    if marks < 0 or marks > 100:
        return "Invalid"
    elif marks >= 90:
        return "A+"
    elif marks >= 80:
        return "A"
    elif marks >= 70:
        return "B"
    elif marks >= 60:
        return "C"
    elif marks >= 50:
        return "D"
    elif marks >= 40:
        return "E"
    else:
        return "F"

print("Student Grade Calculator")
print("------------------------")

for student, marks in student_marks.items():
    print(f"{student}: {marks} marks -> Grade {calculate_grade(marks)}")
```

## Conclusion

The program assigns grades using conditional statements and validates the marks range.
